```python
Wholife_copy = WHO_life.copy()
pib_copy = pib2020_2025.copy()

Wholife_copy = Wholife_copy.rename(columns={'Location':'Country'})

mergeado = pd.merge(Wholife_copy, pib_copy, on= 'Country', how='inner')
```

```python
df_birth = mergeado[mergeado['Indicator'] == 'Life expectancy at birth (years)'].copy()

df_60 = mergeado[mergeado['Indicator'] == 'Life expectancy at age 60 (years)'].copy()

```

```python
pivot = mergeado.pivot_table(index='Country',
                             columns='Indicator',
                             values='Value_num',
                             aggfunc='mean')
pivot['Diff_birth_60'] = pivot['Life expectancy at birth (years)'] - pivot['Life expectancy at age 60 (years)']
pivot.sort_values('Diff_birth_60', ascending=False).head()
```
