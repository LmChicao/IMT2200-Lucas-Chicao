```python
import pandas as pd

adelie = pd.read_csv("adelie.csv")
chinstrap = pd.read_csv("chinstrap.csv")
gentoo = pd.read_csv("gentoo.csv")
```

```python
adelie["species"] = "adelie"
chinstrap["species"] = "chinstrap"
gentoo["species"] = "gentoo"

```

```python
penguins = pd.DataFrame(columns=['island', 'culmen_length_mm', 'culmen_depth_mm', 'flipper_length_mm',
       'body_mass_g', 'sex', "species"])
#penguins = pd.merge(penguins, adelie)
```

```python
penguins = pd.concat([penguins, adelie, chinstrap, gentoo], axis=0, keys=['island', 'culmen_length_mm', 'culmen_depth_mm', 'flipper_length_mm',
       'body_mass_g', 'sex', "species"])
```
