```python
# respuestas
exped = pd.read_csv('data\\exped.csv')
#print('exped_csv',exped.columns) 

'''
peakid, expid, year, season
'''

peaks = pd.read_csv('data\\peaks.csv')
#print('peaks_csv',peaks.columns)
'''
peakid, pkname, pkname2, heightm, pexpid(no se si estara bien), trekyear(nose si estara bien), pyear(nose si estara bien),  
pseason(no se si estara bien), 
'''
refer = pd.read_csv('data\\refer.csv', encoding="latin1")
#print('refer_csv',refer.columns)

'''
expid, ryear(no se si estara bien), 
'''

members = pd.read_csv('data\\members.csv')
#print('members_csv',members.columns)

'''
peakid, expid, myear(no se si estara bien), mseason(no se si estara bien), sex, yob, citizen, msuccess
'''

peaks_with_exp = pd.read_csv('data\\peaks_with_exp.csv')
#print('peaks_with_exp_csv', peaks_with_exp.columns)
'''
peakid, pkname, heightm, success_exp, failed_exp, expid, year, season, memid, sex, yob, citizen, msuccess
'''
print('uwu')
```

```python
df_exped = exped[['expid', 'peakid', 'year', 'season']]
df_peaks = peaks[['peakid', 'pkname', 'heightm']]
df_members = members[['expid', 'peakid', 'sex', 'yob', 'citizen', 'msuccess']]
df_exp_peaks = pd.merge(df_exped, df_peaks, on='peakid', how='left')

df_final = pd.merge(df_exp_peaks, df_members, on=['expid', 'peakid'], how='left')
success_fail = df_members.groupby('peakid')['msuccess'].value_counts().unstack(fill_value=0).reset_index()
success_fail = success_fail.rename(columns={True: 'success_exp', False: 'failed_exp'})


df_final = pd.merge(df_final, success_fail, on='peakid', how='left')
df_final
```

```python
#libreria xd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix, classification_report
```

```python
#5.0)


clean_df['msuccess'] = clean_df['msuccess'].astype(int)
#clean_df['msuccess']

X = clean_df[['age', 'sex', 'citizen', 'season', 'heightm', 'year', 'success_exp', 'failed_exp']]
y = clean_df['msuccess']

```

```python
#5.1)
X_encoded = pd.get_dummies(X, columns=['sex', 'citizen', 'season'], drop_first=True)

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.3, random_state=42)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

'''
StandardScaler funciona bien cuando los atributos numericos están distribuidos de forma aproximadamente normal
'''
```

```python
#5.2)

#KNN
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_scaled, y_train)

'''
n_neighbors=5 (valor por defecto)
'''

#arbolito
tree = DecisionTreeClassifier(max_depth=5, random_state=42)
tree.fit(X_train, y_train)
'''
max_depth=5 para limitar la complejidad del árbol y evitar overfitting.
'''

#Regresion logstica
logreg = LogisticRegression(max_iter=1000)
logreg.fit(X_train_scaled, y_train)
'''
max_iter=1000 garantiza la convergencia y se usa penalización L2 por defecto.
'''
```

```python
#5.2) evaluacion 

y_pred_knn = knn.predict(X_test_scaled)

print("KNN")
print("Accuracy:", accuracy_score(y_test, y_pred_knn))
print("Precisión:", precision_score(y_test, y_pred_knn))
print("Recall:", recall_score(y_test, y_pred_knn))
print("Matriz de confusión:\n", confusion_matrix(y_test, y_pred_knn))
print("------")

y_pred_tree = tree.predict(X_test)  # arbolito no necesita datos escalados
print("Arbolito")
print("Accuracy:", accuracy_score(y_test, y_pred_tree))
print("Precisión:", precision_score(y_test, y_pred_tree))
print("Recall:", recall_score(y_test, y_pred_tree))
print("Matriz de confusión:\n", confusion_matrix(y_test, y_pred_tree))
print("------")

y_pred_rlog = logreg.predict(X_test_scaled)
print("Regresion logística")
print("Accuracy:", accuracy_score(y_test, y_pred_rlog))
print("Precisión:", precision_score(y_test, y_pred_rlog))
print("Recall:", recall_score(y_test, y_pred_rlog))
print("Matriz de confusión:\n", confusion_matrix(y_test, y_pred_rlog))
print("------")
```
