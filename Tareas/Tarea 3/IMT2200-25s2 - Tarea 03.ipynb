```python
# respuestas

niveles = ["Parvularia", "Primer Ciclo Básico", "Segundo Ciclo Básico", "Media"]
dataframios = [df_parvularia, df_ciclo1, df_ciclo2, df_media]

for i in range(4):
    dataframio = dataframios[i]
    promedio_ivm = dataframio.groupby("DS_TIPO_DEPENDENCIA")["IVM Establecimiento"].mean().sort_values()
    plt.figure(figsize=(8, 5))
    promedio_ivm.plot(kind="bar")
    
    plt.title(f"IVM promedio según dependencia ({niveles[i]})")
    plt.xlabel("Tipo de dependencia")
    plt.ylabel("IVM promedio")
    plt.xticks(rotation=45, ha="right")
    plt.show()
```
